## Notebook 06: Business Insights and Recommendations
## Objectives
This notebook summarizes the forecasting results and translates the machine learning findings into actionable business insights. Based on the best-performing forecasting model, recommendations are provided to support inventory planning, demand forecasting, and retail decision-making.

## 1. Forecasting Performance Summary

Five forecasting models were evaluated in this project, ranging from simple baseline methods to advanced machine learning algorithms.

Among all evaluated models, **XGBoost** achieved the highest forecasting accuracy and was selected as the final forecasting model.

| Model | MAE | RMSE |
|----------------------|---------:|---------:|
| Naive Forecast | 1718.51 | 3942.46 |
| Linear Regression | 2019.96 | 3986.24 |
| Ridge Regression | 2027.24 | 3985.36 |
| Random Forest | 1538.27 | 3406.64 |
| **XGBoost** | **1484.80** | **3168.98** |

Compared with the Naive Forecast baseline, XGBoost reduced MAE by approximately **13.6%** and RMSE by approximately **19.6%**, demonstrating a substantial improvement in forecasting accuracy.

## 2. Key Findings

The project identified several important findings regarding Walmart weekly sales forecasting.

- Historical sales were the strongest predictors of future demand.
- The previous week's sales (`lag_1`) contributed the most to forecasting performance.
- The four-week rolling average (`rolling_mean_4`) captured recent demand trends and significantly improved prediction accuracy.
- Linear models were unable to capture the complex relationships within the data.
- Tree-based ensemble models substantially outperformed linear regression approaches.
- XGBoost achieved the best overall performance by effectively modeling nonlinear demand patterns and feature interactions.

## 3. Business Insights

### Inventory Planning

Since recent sales history is the strongest predictor of future demand, inventory replenishment decisions should place greater emphasis on short-term sales trends rather than relying solely on historical averages.

### Promotion Planning

Although promotional variables contributed less than historical demand, they still influenced sales performance. Promotional campaigns should therefore be planned together with demand forecasts to avoid stock shortages or excess inventory.

### Seasonal Demand

Weekly seasonality was more informative than monthly seasonality, suggesting that demand planning should be performed at a weekly level instead of a monthly level.

### Demand Stability

The strong importance of lag features indicates that customer demand is relatively stable over short time horizons, making short-term forecasting particularly effective for inventory management.

## 4. Business Recommendations

Based on the forecasting results, the following recommendations are proposed:

- Use XGBoost as the primary forecasting model for weekly demand prediction.
- Continuously update forecasting models using the latest sales data.
- Monitor rapid changes in demand during holiday periods and promotional events.
- Incorporate forecasting outputs into inventory replenishment and safety stock planning.
- Regularly retrain the model to maintain forecasting accuracy as customer demand patterns evolve.

## 5. Limitations

- The dataset covers only approximately three years of historical sales.
- External factors such as competitor promotions, weather events, and local economic conditions were not fully captured.
- Hyperparameter optimization was limited to maintain reasonable computational time.
- The forecasting models were evaluated using historical data only and were not deployed in a real-time retail environment.

## Conclusion

This project developed an end-to-end demand forecasting pipeline using Walmart weekly sales data, beginning with data understanding and feature engineering, followed by baseline forecasting and machine learning model development.

Among the evaluated models, XGBoost achieved the best forecasting performance, reducing MAE by approximately **13.6%** and RMSE by approximately **19.6%** compared with the Naive Forecast baseline.

The analysis demonstrated that historical demand features, particularly the previous week's sales and recent sales trends, were the primary drivers of forecasting performance. These findings suggest that machine learning models can effectively support retail inventory planning by providing more accurate demand forecasts.
